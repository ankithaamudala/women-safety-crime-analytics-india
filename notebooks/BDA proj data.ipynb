```python
women_crime_keywords = [
    "ASSAULT",
    "SEXUAL",
    "RAPE",
    "KIDNAPPING",
    "HUMAN TRAFFICKING",
    "TRAFFICKING"
]
```

```python
#Filter dataset to keep ONLY required crimes
pattern = "|".join(women_crime_keywords)

df_women = df_6[df_6["OFFENSE"].str.contains(pattern, na=False)]
```

```python
#Remove non-relevant crimes that still contain keywords
exclude_keywords = [
    "THEFT",
    "ROBBERY",
    "BURGLARY",
    "MOTOR",
    "AUTO"
]

exclude_pattern = "|".join(exclude_keywords)

df_women = df_women[~df_women["OFFENSE"].str.contains(exclude_pattern, na=False)]
```

```python
df_women["OFFENSE"].value_counts()
```

```python
#Convert date columns to datetime
date_cols = ["REPORT_DAT", "START_DATE", "END_DATE"]

for col in date_cols:
    df_women[col] = pd.to_datetime(df_women[col], errors="coerce")
```

```python
#Remove rows without location
df_women = df_women.dropna(subset=["LATITUDE", "LONGITUDE"])
```

```python
#Reset index
df_women.reset_index(drop=True, inplace=True)
```

```python
#Extract time features
df_women["hour"] = df_women["START_DATE"].dt.hour
df_women["day"] = df_women["START_DATE"].dt.day
df_women["month"] = df_women["START_DATE"].dt.month
df_women["weekday"] = df_women["START_DATE"].dt.day_name()
```

```python
#Night vs Day classification
def time_of_day(hour):
    if hour >= 20 or hour <= 5:
        return "Night"
    elif 6 <= hour <= 11:
        return "Morning"
    elif 12 <= hour <= 16:
        return "Afternoon"
    else:
        return "Evening"

df_women["time_period"] = df_women["hour"].apply(time_of_day)
```
