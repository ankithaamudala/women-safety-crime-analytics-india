```python
df2 = pd.read_csv(r"D:\Ankitha\2nd Yr\Term - 1\Big Data Analytics\Project\final_ncrb_women_crimes.csv")
df4 = pd.read_csv(r"D:\Ankitha\2nd Yr\Term - 1\Big Data Analytics\Project\final_outdoor_women_crimes_2017_normalized.csv")
```

```python
df2["source_dataset"] = "ncrb_women"
df4["source_dataset"] = "outdoor_2017"
```

```python
df2 = df2[df2["District"].str.upper() != "TOTAL"]
df4 = df4[df4["district"].str.upper() != "TOTAL"]
```

```python
print(df2.columns.tolist())
print(df4.columns.tolist())
```

```python
df2.columns = df2.columns.str.strip().str.lower()
```

```python
df4.columns = df4.columns.str.strip().str.lower()
```

```python
df2_agg = df2[[
    "state",
    "district",
    "year",
    "crime_type",
    "crime_count",
    "source_dataset"
]].copy()

df2_agg["area_type"] = "DISTRICT"
```

```python
df4_agg = df4[[
    "state",
    "district",
    "year",
    "crime_type",
    "crime_count",
    "source_dataset"
]].copy()

df4_agg["area_type"] = "DISTRICT"
```

```python
df2_agg = df2_agg[df2_agg["district"].str.upper() != "TOTAL"]
df4_agg = df4_agg[df4_agg["district"].str.upper() != "TOTAL"]
```

```python
agg_df = pd.concat([df2_agg, df4_agg], ignore_index=True)
```

```python
def calculate_risk(count):
    if count >= 500:
        return "High"
    elif count >= 100:
        return "Medium"
    else:
        return "Low"
```

```python
agg_df["risk_level"] = agg_df["crime_count"].apply(calculate_risk)
```

```python
agg_df[["district", "crime_count", "risk_level"]].head()
```

```python
agg_df["risk_level"].value_counts()

```

```python
agg_df.to_csv("final_aggregated_district_dataset.csv", index=False)
```
