```python
import pandas as pd
import os
```

```python
ALLOWED_CRIMES = [
    "ASSAULT",
    "SEXUAL ASSAULT",
    "KIDNAPPING",
    "RAPE"
]

def standardize_dataset(
    file_path,
    crime_col,
    state_col,
    district_col=None
):
    df = pd.read_csv(file_path)

    # Create District if missing
    if district_col is None:
        df["District"] = "ALL"
        district_col = "District"

    # Detect year columns
    year_cols = [c for c in df.columns if c.isdigit()]

    # Convert wide → long
    df_long = df.melt(
        id_vars=[state_col, district_col, crime_col],
        value_vars=year_cols,
        var_name="Year",
        value_name="Crime_Count"
    )

    # Rename columns
    df_long = df_long.rename(columns={
        state_col: "State",
        district_col: "District",
        crime_col: "Crime_Type"
    })

    # Normalize crime names
    df_long["Crime_Type"] = df_long["Crime_Type"].str.upper().str.strip()

    # FILTER only women-safety outdoor crimes
    df_long = df_long[df_long["Crime_Type"].isin(ALLOWED_CRIMES)]

    # Keep required columns only
    df_final = df_long[
        ["State", "District", "Year", "Crime_Type", "Crime_Count"]
    ]

    # Clean
    df_final = df_final.dropna()
    df_final["Year"] = df_final["Year"].astype(int)
    df_final["Crime_Count"] = df_final["Crime_Count"].astype(int)

    return df_final

```

```python
df1 = standardize_dataset(
    file_path="women-crimedataset-India.csv",
    crime_col="Crime",        # adjust if column name differs
    state_col="State",
    district_col="District"
)

df1.to_csv("cleaned_women_crime.csv", index=False)

```
